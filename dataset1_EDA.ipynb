```python
import pandas as pd
import matplotlib.pyplot as plt

# Path to the CSV file for dataset 1
ds1 = 'C:\\Users\\awrro\\Machine_Learning1\\svc_most_recent_summary_data.csv' # path to the CSV file for dataset 1: 
def load_data(file_path):
    """
    Load data from a CSV file into a pandas DataFrame.

    Parameters:
    file_path (str): The path to the CSV file.

    Returns:
    pd.DataFrame: A DataFrame containing the loaded data.
    """
    try:
        data = pd.read_csv(file_path)
        return data
    except Exception as e:
        print(f"Error loading data: {e}")
        return None           

df1 = load_data(ds1)
# cleaning up DataFrame 1
print(df1.columns)
df1 = df1.drop(columns=['_id']) # removing ID number column from the data and instead rely on pandas built in row counter 

df1 = df1.dropna(subset=['location_name', 'avg_daily_vol','avg_speed', 'avg_weekday_daily_vol' , 'avg_weekend_daily_vol', 'avg_wkdy_am_peak_start', 
                         'avg_wkdy_am_peak_vol', 'avg_wkdy_pm_peak_start', 'avg_wkdy_pm_peak_vol']) # removing any rows with missing values in the specified columns

df1 = df1[['location_name', 'avg_daily_vol', 'avg_weekday_daily_vol' , 'avg_speed', 'avg_weekend_daily_vol', 'avg_wkdy_am_peak_start', 'avg_wkdy_am_peak_vol', 
            'avg_wkdy_pm_peak_start', 'avg_wkdy_pm_peak_vol']] # taking only the relevant columns for the second milestone

# viewing the first 50 rows of the cleaned DataFrame
print(df1.head(25))

df1.plot(y='avg_daily_vol', x='avg_wkdy_pm_peak_vol', kind='scatter', title='Scatter Plot of Average Daily Volume vs Average Weekday PM Peak Volume', xlabel='Average Daily Volume', ylabel='Average Weekday PM Peak Volume')
df1.plot(y='avg_daily_vol', x='avg_wkdy_am_peak_vol', kind='scatter', title='Scatter Plot of Average Daily Volume vs Average Weekday AM Peak Volume', ylabel='Average Daily Volume', xlabel='Average Weekday AM Peak Volume')
df1.plot(x='avg_speed', y='avg_daily_vol', kind='scatter', title='Scatter Plot of Average Speed vs Average Daily Volume', xlabel='Average Speed', ylabel='Average Daily Volume')
```
